```python
# STEP 5: INITIAL DATA CLEANING

# Create a copy of the original dataset
df_clean = df.copy()

# Remove leading and trailing spaces from column names
df_clean.columns = df_clean.columns.str.strip()

# Remove leading and trailing spaces from text columns
for col in df_clean.select_dtypes(include='object').columns:
    df_clean[col] = df_clean[col].str.strip()

# Check duplicate customer IDs
print("Duplicate Customer IDs:",
      df_clean['CustomerID'].duplicated().sum())

# Check missing values again
print("\nMissing Values:")
print(df_clean.isnull().sum())

# Display cleaned dataset
print("\nCleaned Dataset Shape:", df_clean.shape)

display(df_clean.head())
```

```python
# STEP 7: INVESTIGATE MISSING VALUES

# 1. Check churn distribution
print("Churn Label Distribution:")
print(df_clean['ChurnLabel'].value_counts(dropna=False))

print("\nChurn Percentage:")
print(
    (df_clean['ChurnLabel'].value_counts(normalize=True) * 100).round(2)
)

# 2. Check relationship between CustomerStatus and ChurnLabel
print("\nCustomer Status vs Churn Label:")
print(pd.crosstab(df_clean['CustomerStatus'], df_clean['ChurnLabel']))

# 3. Check missing Offer values
print("\nOffer Missing Values by Churn:")
print(
    df_clean.groupby('ChurnLabel')['Offer']
    .apply(lambda x: x.isna().sum())
)

# 4. Check missing InternetType values against InternetService
print("\nInternet Service vs Internet Type:")
print(pd.crosstab(
    df_clean['InternetService'],
    df_clean['InternetType'],
    dropna=False
))

# 5. Check missing values in important columns
print("\nMissing Value Summary:")
print(df_clean[['Offer', 'InternetType',
                'ChurnCategory', 'ChurnReason']].isnull().sum())
```

```python
# STEP 9: HANDLE MISSING VALUES

df_model = df_clean.copy()

# 1. Replace missing Offer values
df_model['Offer'] = df_model['Offer'].fillna('No Offer')

# 2. Replace missing InternetType values
# Missing InternetType corresponds to customers without InternetService
df_model['InternetType'] = df_model['InternetType'].fillna('No Internet')

# 3. Remove columns that reveal the churn outcome
columns_to_drop = [
    'CustomerID',
    'CustomerStatus',
    'ChurnScore',
    'ChurnCategory',
    'ChurnReason'
]

df_model = df_model.drop(columns=columns_to_drop)

# 4. Separate target variable
y = df_model['ChurnLabel'].map({
    'No': 0,
    'Yes': 1
})

# 5. Remove target from input features
X = df_model.drop(columns=['ChurnLabel'])

# 6. Check the results
print("Features Shape:", X.shape)
print("Target Shape:", y.shape)

print("\nRemaining Missing Values:")
print(X.isnull().sum()[X.isnull().sum() > 0])

print("\nTarget Distribution:")
print(y.value_counts())

print("\nOffer Categories:")
print(X['Offer'].value_counts())

print("\nInternet Type Categories:")
print(X['InternetType'].value_counts())
```

```python
# STEP 10: FINAL DATA QUALITY CHECK

print("Input Features:", X.shape)
print("Target:", y.shape)

print("\nMissing values in X:", X.isnull().sum().sum())
print("Missing values in y:", y.isnull().sum())

print("\nData Types:")
print(X.dtypes.value_counts())

print("\nTarget Values:")
print(y.value_counts())

print("\nTarget Percentage:")
print((y.value_counts(normalize=True) * 100).round(2))
```

```python
# STEP 14: IDENTIFY FEATURE TYPES

numeric_features = X.select_dtypes(
    include=['int64', 'float64']
).columns.tolist()

categorical_features = X.select_dtypes(
    include=['object', 'category', 'bool']
).columns.tolist()

print("Numerical Features:", len(numeric_features))
print(numeric_features)

print("\nCategorical Features:", len(categorical_features))
print(categorical_features)
```

```python
# STEP 15: CHECK IMPORTANT FEATURES

important_columns = [
    'Age',
    'TenureinMonths',
    'MonthlyCharge',
    'TotalCharges',
    'TotalRevenue',
    'TotalRefunds',
    'SatisfactionScore',
    'CLTV',
    'Number_of_Referrals',
    'AvgMonthlyGBDownload'
]

for col in important_columns:
    if col in X.columns:
        print(f"\n--- {col} ---")
        print(
            X.groupby(y)[col]
            .agg(['mean', 'median', 'min', 'max'])
            .round(2)
        )
```
